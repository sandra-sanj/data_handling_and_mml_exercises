### # Vaihe 1: Voiko onnettomuuden vakavuuden ennustaa liikenneolosuhteiden perusteella?

```python
# Vaihe 1: Voiko onnettomuuden vakavuuden ennustaa liikenneolosuhteiden perusteella?
df = df_original
```

```python
# df = df.loc[:, [kolumnien nimet]]
```

```python
# 0 = Talvi(tammikuu/1, helmuikuu/2, joulukuu/12), 1 = Kevät(maaliskuu/3, huhtikuu/4, toukokuu/5), 2 = Kesä(kesäkuu/6, heinäkuu/7, elokuu/8), 3 = Syksy(syyskuu/9, lokakuu/10, marraskuu/11)
df['Kuukausi'] = df['Kuukausi'].replace({1 : 0, 2 : 0, 3 : 1, 4 : 1, 5 : 1, 6 : 2, 7 : 2, 8 : 2, 9 : 3, 10 : 3, 11 : 3, 12 : 0})
df = df.rename(columns={'Kuukausi' : "Vuodenaika"})
df
```

```python

df['Valoisa'] = df['Valoisuus'].map({'tie valaistu' : 1, 'pimeä (valaisematon)' : 0, 'päivänvalo' : 1, 'hämärä' : 0})
```

```python
df['Taajama'] = df['Taajama'].map({'ei': 0, 'kyllä': 1})
```

```python
df['Raskaan liikenteen onnettomuus'] = df['Raskaan liikenteen onnettomuus'].map({'ei': 0, 'kyllä': 1})
```

```python
df['Hyvä_sää'] = df['Sää'].map({'kirkas' : 1, 'lumisade' : 0, 'pilvipouta' : 1, 'raesade' : 0, 'vesisade' : 0, 'räntäsade' : 0,
 'sumu' : 0})
```

```python
df['Kestopäällyste'] = df['Päällyste'].map({'kestopäällyste' : 1, 'sora' : 0, 'kivi' : 0, 'öljysora tai vast' : 0, 'muu' : 0, 'betoni' : 1})
```

```python
# kuiva=0, märkä=1, luminen=2
df['Pinta'] = df['Pinta'].map({'jäinen' : 2, 'luminen' : 2, 'urissa vettä' : 1,
   'paljas, märkä' : 1, 'ajourat paljaat' : 2, 'paljas, kuiva' : 0, 'sohjoinen' : 2})
```

```python
# Ensimmäinen osuus
df.drop(columns=['OnnettomuusID', 'Tienpitäjä', 'Tieosoite', 'Ajorata', 'Tie', 'Osa', 'Etaisyys', 'ELY-keskus', 'Maakunta', 'Kunta', 'Katuosoite', 'Vuosi', 'Onnettomyystyyppi', 'Onnettomuustyyppi nro', 'Onnettomuusluokka', 'Tietyö', 'Onnettomuuspaikka', 'Liittymäjärjestelyt', 'Viikonpäivä', 'Verkollinen asema', 'Talvihoitoluokitus', 'Toiminallinen luokka korjattu', 'Rautatie', 'Liikennevalot', 'Valoisuus', 'Sää', 'Päällyste', 'x-koordinaatti', 'y-koordinaatti'], inplace=True)
df
```

```python
df.dropna(how='any', inplace=True)
df
```

```python
df.dtypes
```

```python
df['Lämpötila'] = (df['Lämpötila'].str.replace('−', '-', regex=False).astype(float)) # Korjataan "väärä" miinusmerkki
df
```

```python
df['Vakavuus'] = df['Vakavuus'].map({'Ei henkilövahinkoja': 0, 'Loukkaantumiseen johtanut': 1, 'Kuolemaan johtanut' : 2, 'Vakavaan loukkaantumiseen johtanut' : 1})
df
```

```python
# one-hot encode the genetic_variant variable
df = pd.get_dummies(df, columns=['Pinta', 'Vuodenaika'], dtype='int64')
df
```

```python
import matplotlib.pyplot as plt

def plot_data(df):
    plt.figure(figsize=(12, 25))
    for i, col in enumerate(df.select_dtypes(include=['int', 'float'])):
        plt.subplot(10, 4, i + 1)
        df[col].plot(kind='hist', title=col)
        plt.xlabel(col)
        plt.subplots_adjust(hspace=0.5)
```

```python
plot_data(df)
```

```python
# Tasapainotusta

kuolema = df[df['Vakavuus'] == 2]
kuolema
```

```python
ei_henkilovahinkoja = df[df['Vakavuus'] == 0].sample(n=300,random_state=42)
ei_henkilovahinkoja 
```

```python
loukkaantuminen = df[df['Vakavuus'] == 1].sample(n=300,random_state=42)
loukkaantuminen
```

```python
balanced_df = pd.concat([kuolema, loukkaantuminen, ei_henkilovahinkoja])
balanced_df
```

```python
plot_data(balanced_df)
```

```python
# Modeling

balanced_targets = balanced_df['Vakavuus']
balanced_df = balanced_df.drop(columns=['Vakavuus'])
balanced_df
```

```python
from sklearn.metrics import confusion_matrix
from sklearn.model_selection import cross_val_score, cross_val_predict
from sklearn.metrics import classification_report
from sklearn.ensemble import RandomForestClassifier

# Create a random forest classifier
model = RandomForestClassifier(n_estimators=100, bootstrap=True, max_samples=0.75, max_features=0.65, random_state=123)

model.fit(balanced_df, balanced_targets)

# Carry out 10-fold cross-validation
target_pred = cross_val_predict(estimator=model, X=balanced_df, y=balanced_targets, cv=10)

print(f'Confusion matrix:\n{confusion_matrix(balanced_targets, target_pred)}')
print(classification_report(balanced_targets, target_pred))

# Display the feature importances
for value, name in zip(model.feature_importances_, model.feature_names_in_):
    print(name + ":  " + str(value))
```
